```python
# Storing NEO url
NEO_ENDPOINT = "https://api.nasa.gov/neo/rest/v1/feed"
```

```python
# Setting up parameters for the request
parameters = {"start_date": "",
               "api_key": API_KEY}
```

```python
# Sends a request and returns a status code
neo_response = requests.get(NEO_ENDPOINT, params = parameters)
neo_response.status_code
```

```python
# returns the response as a json format or dictionary of near earth objects
neo_data = neo_response.json() 

# extracts the near earth objects from the data
neo_obj = neo_data["near_earth_objects"] 

# returns data near the earth on 25 june (7 days from the start date)
next_obj_date = neo_obj["2026-06-25"]

# creates an empty list
object_list = []

# Loops through list of dictionaries
for i in next_obj_date:
    close_approach = i["close_approach_data"] #returns a list of dictionaries holding close_approach data
    
    for s in close_approach:  #loops through the close approach data and returns the relative velocity
        rel_velocity =  s["relative_velocity"]["kilometers_per_second"]
        
    # creates a list of dictionaries for each individual object
    obj = {"id":i["id"],
           "name":i["name"],
          "est_distance_km": i["estimated_diameter"]["kilometers"]["estimated_diameter_max"],
           "abs_mag": i["absolute_magnitude_h"],
           "relative_velocity_km/s": rel_velocity 
          }
    # appends every dictionary item to the object list
    object_list.append(obj)
object_list
```

```python
# creates a dataframe using pandas library
neo_df = pd.DataFrame(object_list)
neo_df
```

```python
#converts dataframe into a shareable csv file
neo_df.to_csv('NEO_DATA.csv', index=False)
```
